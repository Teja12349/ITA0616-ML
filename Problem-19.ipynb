```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

data = {
    'Income': [20000, 30000, 40000, 50000, 60000, 70000],
    'Age': [22, 25, 28, 35, 40, 45],
    'Loan': ['No', 'No', 'Yes', 'Yes', 'Yes', 'Yes']
}

df = pd.DataFrame(data)

X = df[['Income', 'Age']]
y = LabelEncoder().fit_transform(df['Loan'])

model = GaussianNB()
model.fit(X, y)

prediction = model.predict([[55000, 32]])

result = 'Yes' if prediction[0] == 1 else 'No'

print("Bank Loan Approval:", result)
```
