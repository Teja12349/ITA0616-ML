```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

data = {
    'RAM': [2, 3, 4, 4, 6, 8, 8, 12],
    'Battery': [2500, 3000, 3500, 4000, 4500, 5000, 5500, 6000],
    'Storage': [32, 32, 64, 64, 128, 128, 256, 256],
    'Price_Range': [0, 0, 1, 1, 2, 2, 3, 3]
}

df = pd.DataFrame(data)

X = df[['RAM', 'Battery', 'Storage']]
y = df['Price_Range']

model = RandomForestClassifier(random_state=42)
model.fit(X, y)

prediction = model.predict([[8, 5000, 128]])

print("Predicted Mobile Price Range:", prediction[0])
```
