```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

data = {
    'Income': [20000, 30000, 50000, 60000, 80000, 90000],
    'Loan': [15000, 20000, 10000, 15000, 20000, 10000],
    'Age': [22, 25, 30, 35, 40, 45],
    'Credit_Score': ['Poor', 'Poor', 'Average', 'Average', 'Good', 'Good']
}

df = pd.DataFrame(data)

X = df[['Income', 'Loan', 'Age']]
y = df['Credit_Score']

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

prediction = model.predict([[70000, 12000, 32]])

print("Credit Score:", prediction[0])
```
